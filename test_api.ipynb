```python
transactions = transactions_raw[['id', 'description', 'amount', 'currencyCode', 'date', 'category', 'type', 'status', 'operationType', 'createdAt']]
# transactions['bankName'] = 'NuBank'
transactions.head()
```

```python
merchants = pd.json_normalize(transactions_raw.loc[~transactions_raw['merchant'].isnull()][['merchant']].drop_duplicates()['merchant'])
merchants = merchants.rename(columns={'category': 'merchantCategory', 'cnae': 'merchantCnae', 'cnpj': 'merchantCnpj', 'id': 'merchantId', 'businessName': 'merchantName'})

merchants.head(5)
```

```python
transactions_final = transactions.join(merchants, how='left')
transactions_final.head(5)
```
